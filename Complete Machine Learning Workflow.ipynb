```python
# ==========================
# 11. Evaluate Model
# ==========================

# ==========================================
# 11.1. Regression Evaluation Metrics
# ==========================================
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np
import pandas as pd

# Generate predictions on the test set
y_pred = model.predict(X_test, verbose=0)

# ------------------------------------------
# 11.1.1. MAE (Mean Absolute Error)
# Average absolute difference between
# actual and predicted values.
#
# Lower is better.
# Same units as the target variable.
# Less sensitive to outliers than MSE/RMSE.
# ------------------------------------------
mae = mean_absolute_error(y_test, y_pred)

# ------------------------------------------
# 11.1.2. MSE (Mean Squared Error)
# Average squared difference between
# actual and predicted values.
#
# Lower is better.
# Penalizes large errors heavily.
# Common loss function for neural networks.
# ------------------------------------------
mse = mean_squared_error(y_test, y_pred)

# ------------------------------------------
# 11.1.3. RMSE (Root Mean Squared Error)
# Square root of MSE.
#
# Lower is better.
# Same units as target variable.
# Often easier to interpret than MSE.
# ------------------------------------------
rmse = np.sqrt(mse)

# ------------------------------------------
# 11.1.4. R² Score (Coefficient of Determination)
#
# Measures how much variance in the target
# variable is explained by the model.
#
# Interpretation:
# 1.0  = Perfect prediction
# 0.0  = Same as predicting the mean
# < 0  = Worse than predicting the mean
# ------------------------------------------
r2 = r2_score(y_test, y_pred)

# Create summary table
regression_results = pd.DataFrame({
    'Metric': ['MAE', 'MSE', 'RMSE', 'R²'],
    'Value': [mae, mse, rmse, r2]
})

print(regression_results)

# ==========================================
# 11.2. Classification Evaluation Metrics
# ==========================================

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
import pandas as pd

# Generate class predictions
y_pred = model.predict(X_test)

# ------------------------------------------
# Confusion Matrix
#
# TN = True Negative
# FP = False Positive
# FN = False Negative
# TP = True Positive
#
# Provides detailed insight into
# classification performance.
# ------------------------------------------
cm = confusion_matrix(y_test, y_pred)

# ------------------------------------------
# Accuracy
#
# Percentage of correctly classified samples.
#
# Best used when classes are balanced.
# ------------------------------------------
accuracy = accuracy_score(y_test, y_pred)

# ------------------------------------------
# Precision
#
# Of all predicted positives,
# how many were actually positive?
#
# Important when False Positives
# are costly.
# ------------------------------------------
precision = precision_score(y_test, y_pred)

# ------------------------------------------
# Recall (Sensitivity)
#
# Of all actual positives,
# how many did the model detect?
#
# Important when False Negatives
# are costly.
# ------------------------------------------
recall = recall_score(y_test, y_pred)

# ------------------------------------------
# F1 Score
#
# Harmonic mean of Precision and Recall.
#
# Useful when classes are imbalanced.
# ------------------------------------------
f1 = f1_score(y_test, y_pred)

# Generate prediction probabilities
# Required for ROC-AUC
#
# NOTE:
# For many sklearn models:
# y_prob = model.predict_proba(X_test)[:,1]
#
# For neural networks:
# y_prob = model.predict(X_test).ravel()
#
# Example below assumes binary classification.
y_prob = model.predict_proba(X_test)[:,1]

# ------------------------------------------
# ROC-AUC
#
# Measures the model's ability to
# distinguish between classes.
#
# Interpretation:
# 1.0 = Perfect
# 0.5 = Random guessing
# ------------------------------------------
roc_auc = roc_auc_score(y_test, y_prob)

# Create summary table
classification_results = pd.DataFrame({
    'Metric': [
        'Accuracy',
        'Precision',
        'Recall',
        'F1 Score',
        'ROC-AUC'
    ],
    'Value': [
        accuracy,
        precision,
        recall,
        f1,
        roc_auc
    ]
})

print("Confusion Matrix:")
print(cm)

print("\nClassification Metrics:")
print(classification_results)

# ==========================================
# 11.3. Loss Function for ANNs
# ==========================================
# Training loss
loss = history.history['loss']
# Validation loss
val_loss = history.history['val_loss']

plt.plot(loss, label='Training Loss')
plt.plot(val_loss, label='Validation Loss')
plt.legend()
```
